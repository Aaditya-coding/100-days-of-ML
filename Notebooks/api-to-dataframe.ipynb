```python
import time
import requests
import pandas as pd

# Initialize empty list to store dataframes
df_list = []

for i in range(1, 429):
    try:
        # Make API request
        response = requests.get('https://api.themoviedb.org/3/movie/top_rated?api_key=12740bb6c4e336dbaf35cbfc170076dd&language=en-US&page={}'.format(i))
        
        # Check if request was successful
        response.raise_for_status()
        
        # Create temporary dataframe and append to list
        temp_df = pd.DataFrame(response.json()['results'])[['id','title','overview','release_date','popularity','vote_average','vote_count']]
        df_list.append(temp_df)  # Append to list instead of concatenating immediately
        
        # Add delay to avoid rate limiting (adjust as needed)
        time.sleep(0.5)  # 100ms delay between requests
        
        # Optional: Print progress
        if i % 50 == 0:
            print(f"Processed {i} pages...")
            
    except requests.exceptions.RequestException as e:
        print(f"Error on page {i}: {e}")
        # Add longer delay on error before continuing
        time.sleep(1)
        continue

# Concatenate all dataframes at once (more efficient)
if df_list:
    df = pd.concat(df_list, ignore_index=True)
    print(f"Final dataframe shape: {df.shape}")
else:
    print("No data collected")
```
