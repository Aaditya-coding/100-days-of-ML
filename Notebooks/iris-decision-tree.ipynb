## 2. Regression

```python
import pandas as pd
from sklearn import tree
from dtreeviz import model

# Load dataset
boston = pd.read_csv("BostonHousing.csv")

# Features and target
X_train = boston.drop("medv", axis=1)
y_train = boston["medv"]

# Train model
regr = tree.DecisionTreeRegressor(max_depth=1)
regr.fit(X_train, y_train)

# Create visualization model
viz_model = model(
    regr,
    X_train=X_train,
    y_train=y_train,
    feature_names=X_train.columns,
    target_name="price"
)

# Display tree
viz_model.view()
```

## 3. Horizontal Decision Tree

```python
from sklearn.datasets import load_iris
from sklearn import tree
from dtreeviz import model

iris = load_iris()

X_train = iris.data
y_train = iris.target

clas = tree.DecisionTreeClassifier()
clas.fit(X_train, y_train)

m = model(
    clas,
    X_train=X_train,
    y_train=y_train,
    feature_names=iris.feature_names,
    target_name="species",
    class_names=["setosa", "versicolor", "virginica"]
)

m.view(scale=1.5)
```

## 4. Show prediction path

```python
import numpy as np
from sklearn import tree
from sklearn.datasets import load_iris
from dtreeviz import model

# Load data
iris = load_iris()

X_train = iris.data
y_train = iris.target

# Train model
clas = tree.DecisionTreeClassifier()
clas.fit(X_train, y_train)

# Random sample to highlight
X = iris.data[np.random.randint(0, len(iris.data))]

# Create visualization
m = model(
    clas,
    X_train=X_train,
    y_train=y_train,
    feature_names=iris.feature_names,
    target_name="species",
    class_names=["setosa", "versicolor", "virginica"]
)

# Display tree and highlight X
m.view(x=X)
```

## 7. Show just prediction path, nothing else

```python
import numpy as np
from sklearn import tree
from sklearn.datasets import load_iris
from dtreeviz import model

# Load dataset
iris = load_iris()

X_train = iris.data
y_train = iris.target

# Train classifier
clas = tree.DecisionTreeClassifier()
clas.fit(X_train, y_train)

# Random sample to highlight
X = iris.data[np.random.randint(0, len(iris.data))]

# Create visualization model
m = model(
    clas,
    X_train=X_train,
    y_train=y_train,
    feature_names=iris.feature_names,
    target_name="species",
    class_names=["setosa", "versicolor", "virginica"]
)

# Display only the decision path
m.view(
    x=X,
    show_just_path=True
)
```
